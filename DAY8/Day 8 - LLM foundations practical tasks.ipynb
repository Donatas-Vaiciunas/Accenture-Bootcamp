#### 2. Call the OpenAI API with different models

Prompt `"Explain the concept of machine learning to a 10-year-old."`

`model`: `gpt-5.6-luna` and `gpt-5.4-mini`.

Compare the outputs from each model. 

You can evaluate the differences in:
* Clarity of explanation.
* Simplicity of language (suitable for a 10-year-old).
* Creativity in the response.
* Response time.

````python
import time

PROMPT = "Explain the concept of machine learning to a 10-year-old."


def run_model(model):
    print(f"{'='*60}")
    print(f"MODEL: {model}")
    print(f"{'='*60}")
    started = time.perf_counter()
    response = client.chat.completions.create(
        model=model,
        messages=[{"role": "user", "content": PROMPT}],
        max_completion_tokens=2000,
    )
    elapsed = time.perf_counter() - started
    print(response.choices[0].message.content.strip())
    print(f"\n  output tokens: {response.usage.completion_tokens}")
    print(f"  time: {elapsed:.1f}s")


run_model("gpt-5.6-luna")
````

````python
run_model("gpt-5.4-mini")
````

**Observations:**
- **Clarity:** both models explain machine learning as a computer learning from examples instead of following rules written by a person. gpt-5.6-luna builds the explanation step by step and ends with a short recap, so it is a bit easier to follow.
- **Simplicity of language:** gpt-5.4-mini keeps sentences short and uses everyday words. gpt-5.6-luna writes a longer answer and sometimes uses words like "patterns" or "data" that it then has to explain.
- **Creativity:** gpt-5.6-luna uses richer analogies (teaching a pet, learning to recognise animals from pictures). gpt-5.4-mini sticks to one common example.
- **Response time:** gpt-5.4-mini answers noticeably faster. gpt-5.6-luna spends extra time on hidden reasoning before it writes the answer.

**Key insight:** the bigger model gives a more polished and creative explanation, the smaller one is faster and cheaper. For a simple explanation like this, gpt-5.4-mini is good enough.

#### 3. Call the OpenAI API with different `temperature` values

Prompt: `"Write a two sentence story about a robot learning to cook."`

`temperature` values: `0`, `0.7` and `1.5`. Generate three responses at each temperature.

Use `gpt-5.4-mini` for this task and set `reasoning_effort="none"`.

Compare the outputs.

Tip: With Chat Completions API you can use parameter `n=3` to generate three different responses to your prompt with one call.

````python
MODEL = "gpt-5.4-mini"
PROMPT = "Write a two sentence story about a robot learning to cook."


def run_temperature(temp):
    print(f"{'='*60}")
    print(f"TEMPERATURE = {temp}")
    print(f"{'='*60}")
    response = client.chat.completions.create(
        model=MODEL,
        messages=[{"role": "user", "content": PROMPT}],
        temperature=temp,
        n=3,
        max_completion_tokens=100,
        reasoning_effort="none",
    )
    for run, choice in enumerate(response.choices, start=1):
        print(f"  Run {run}: {choice.message.content.strip()}")


run_temperature(0)
````

````python
run_temperature(0.7)
````

````python
run_temperature(1.5)
````

**Observations:**
- **temperature = 0:** the three runs are almost identical: the same robot, the same dish and nearly the same wording. The model always picks the most likely next token.
- **temperature = 0.7:** the stories keep the same idea, but details change between runs: robot names, dishes and endings are different. All stories are still coherent.
- **temperature = 1.5:** every run is different and more unusual. The wording gets strange, and some sentences become hard to follow or slightly broken.

**Key insight:** temperature controls randomness, not quality. Use a low value for repeatable and factual tasks, around 0.7 for creative writing. Very high values trade coherence for variety.

#### 4. Call the OpenAI API with different `max_output_tokens`/`max_completion_tokens` values.

Prompt: `"Explain the concept of machine learning in simple terms."`

`max_output_tokens` (Response API call)/`max_completion_tokens` (Chat Completions API call) values: `50`, `100` and `200`.

Use `gpt-5.4-mini` for this task.

Compare the length and detail of the responses.

````python
PROMPT = "Explain the concept of machine learning in simple terms."


def run_max_tokens(limit):
    print(f"{'='*60}")
    print(f"MAX_COMPLETION_TOKENS = {limit}")
    print(f"{'='*60}")
    response = client.chat.completions.create(
        model=MODEL,
        messages=[{"role": "user", "content": PROMPT}],
        max_completion_tokens=limit,
        reasoning_effort="none",
    )
    print(response.choices[0].message.content.strip())
    print(f"\n  output tokens: {response.usage.completion_tokens}")
    print(f"  finish reason: {response.choices[0].finish_reason}")


run_max_tokens(50)
````

````python
run_max_tokens(100)
````

````python
run_max_tokens(200)
````

**Observations:**
- **50 tokens:** only the definition fits, and the answer is cut off in the middle of a sentence. Finish reason is `length`.
- **100 tokens:** the definition and one example fit, but the answer can still stop before the end.
- **200 tokens:** the answer is complete, with the definition, an everyday example and how the model improves with more data. Finish reason is `stop` when the model finishes on its own.

**Key insight:** `max_completion_tokens` is a hard cut-off, not a length instruction. The model does not plan the answer to fit the limit, it just stops. To get a short answer, ask for it in the prompt.

#### 5. Call the OpenAI API with different `top_p` values.

Prompt: `"Write a list of 5 completely random and unusual hobbies that no one has ever heard of."`

`top_p` values: `0.3`, `0.7`, and `1.0`.

Use `gpt-5.4-mini` for this task and set `reasoning_effort="none"`.

Compare response variation as you adjust `top_p`.

````python
PROMPT = "Write a list of 5 completely random and unusual hobbies that no one has ever heard of."


def run_top_p(top_p):
    print(f"{'='*60}")
    print(f"TOP_P = {top_p}")
    print(f"{'='*60}")
    response = client.chat.completions.create(
        model=MODEL,
        messages=[{"role": "user", "content": PROMPT}],
        top_p=top_p,
        n=3,
        max_completion_tokens=300,
        reasoning_effort="none",
    )
    for run, choice in enumerate(response.choices, start=1):
        print(f"--- Run {run} ---")
        print(choice.message.content.strip())


run_top_p(0.3)
````

````python
run_top_p(0.7)
````

````python
run_top_p(1.0)
````

**Observations:**
- **top_p = 0.3:** the three runs give very similar lists, often with the same hobbies and the same wording. The model samples only from a small set of the most likely tokens.
- **top_p = 0.7:** some hobbies repeat, but more new ideas and different descriptions appear between runs.
- **top_p = 1.0:** the lists differ the most between runs and the hobbies are the most inventive.

**Key insight:** `top_p` limits the pool of tokens the model can choose from. Lower values give safer and more repetitive answers, higher values give more diverse ones. Change either `temperature` or `top_p`, not both.

#### 6. Call the OpenAI API with different `frequency_penalty` values.

Prompt: `"Write a list of funny one-liner jokes about cats."`

`frequency_penalty` values: `0`, `0.5`, and `1.0` (not supported by Responses API).

Use `gpt-5.4-mini` for this task and set `reasoning_effort="none"`.

Compare how the repetition of joke formats changes as you increase the `frequency_penalty`.

````python
PROMPT = "Write a list of funny one-liner jokes about cats."


def run_frequency_penalty(penalty):
    print(f"{'='*60}")
    print(f"FREQUENCY_PENALTY = {penalty}")
    print(f"{'='*60}")
    response = client.chat.completions.create(
        model=MODEL,
        messages=[{"role": "user", "content": PROMPT}],
        frequency_penalty=penalty,
        max_completion_tokens=400,
        reasoning_effort="none",
    )
    print(response.choices[0].message.content.strip())


run_frequency_penalty(0)
````

````python
run_frequency_penalty(0.5)
````

````python
run_frequency_penalty(1.0)
````

**Observations:**
- **frequency_penalty = 0:** many jokes follow the same template ("Why did the cat...?", "My cat..."), and the words "cat" and "because" repeat in almost every line.
- **frequency_penalty = 0.5:** the openings are more varied and fewer jokes use the same structure.
- **frequency_penalty = 1.0:** joke formats vary the most. The model avoids repeating words and even replaces "cat" with "kitty" or "feline".

**Key insight:** `frequency_penalty` lowers the probability of a token the more often it has already appeared, so it reduces repeated words and joke templates. A value that is too high can push the model away from words it actually needs.

Review: "The product is amazing, and I love using it every day!"

Use it in your API call.

Observe how the model's performance changes and compare the response to Zero-Shot output.

````python
EXAMPLES = [
    ("The battery died after two days and support never replied.", "NEGATIVE"),
    ("It works as described, nothing special.", "NEUTRAL"),
    ("Setup took five minutes and it runs perfectly.", "POSITIVE"),
]

messages = [{"role": "system", "content": SYSTEM}]
for example_review, sentiment in EXAMPLES:
    messages.append({"role": "user", "content": example_review})
    messages.append({"role": "assistant", "content": sentiment})
messages.append({"role": "user", "content": REVIEW})

response = client.chat.completions.create(
    model=MODEL,
    messages=messages,
    max_completion_tokens=200,
    reasoning_effort="none",
)

print("=== FEW-SHOT (3 examples) ===")
print(response.choices[0].message.content.strip())
````

- **Few-shot:** the same correct answer, but in exactly the format of the examples: `POSITIVE`, one word in capital letters, nothing else.

**Key insight:** for an easy review the model already knows the sentiment, so the examples did not change the accuracy. The examples taught the FORMAT and the LABEL SET, which makes the answer easy to use in code.

"Generative AI, sometimes called "gen AI", refers to deep learning models that can create complex original content such as long-form text, high-quality images, realistic video or audio and more in response to a user’s prompt or request.

#### 22. Generate a paragraph about a cat without using the letter 'e'/'E'.

Write a prompt that forces the AI to produce a three-sentence story about a cat while strictly forbidding it to use letter 'e'.

Use `gpt-5.4-mini` for this task.

Explain why your prompt is effective.

````python
NO_E_PROMPT = """Write a three-sentence story about a cat.

**Strict rule: the letter 'e' (and 'E') must not appear anywhere in your response.** Not in a single word, not in a name, not in a title.

Follow these steps:
1. Common words like "the", "he", "she", "her", "they", "were", "be", "have" and "very" contain 'e' and are forbidden. Use words like "a", "this", "that", "his", "it", "was" and "a lot" instead.
2. Call the cat by a name without 'e', for example "Tom".
3. Before you answer, check every word letter by letter and replace any word that contains 'e'.

Output only the three sentences, nothing else."""

story = ask(NO_E_PROMPT, max_completion_tokens=8000, reasoning_effort="high")
e_count = story.lower().count("e")
print(story.replace("e", "[E]").replace("E", "[E]") if e_count else story)
print(f"\nLetter 'e' count: {e_count}")
````

**Why the prompt is effective:**
- **The rule is bold and explicit:** it says "e" and "E" are forbidden everywhere, including names and titles, so there is no loophole.
- **It lists forbidden common words with replacements:** the model writes in tokens, not letters, and cannot "see" that the token "the" contains an "e". Naming these words and giving safe alternatives removes the most frequent mistakes.
- **It gives a safe name for the cat:** "Tom" avoids a typical mistake in names.
- **It asks for a self-check:** together with `reasoning_effort="high"`, the model can draft the story, check each word and rewrite the ones that fail before answering.
- **It asks only for three sentences:** less text means fewer chances to break the rule.

**Key insight:** generation happens at the TOKEN level, not the CHARACTER level. A character constraint works only when the prompt turns it into word-level rules and gives the model room to check its own answer.

Which method yields a more reliable result for a computer program to process, and why?

````python
TRIP_TEXT = "Our trip to Tokyo starts on June 12th. We're staying for 5 nights at the Park Hyatt before heading to Kyoto."

print("=== METHOD 1: Prompt-Only ===")
print("(Ask nicely for JSON — no guarantees)\n")
response = client.chat.completions.create(
    model=MODEL,
    messages=[
        {"role": "system", "content": "Extract travel info as JSON with fields: destination, start_date, stay_duration"},
        {"role": "user", "content": TRIP_TEXT},
    ],
    reasoning_effort="none",
)
print(response.choices[0].message.content)
````

````python
from pydantic import BaseModel, Field

print("=== METHOD 2: Schema-Enforced ===")
print("(Pydantic model + parse — guaranteed structure)\n")


class TravelInfo(BaseModel):
    destination: str = Field(description="The first city of the trip")
    start_date: str = Field(description="Start date of the trip, formatted as YYYY-MM-DD")
    stay_duration: int = Field(description="Number of nights as an integer")


response = client.responses.parse(
    model=MODEL,
    instructions="Extract travel information from the text.",
    input=TRIP_TEXT,
    text_format=TravelInfo,
)

travel = response.output_parsed
print(f"Destination:    {travel.destination}")
print(f"Start date:     {travel.start_date}")
print(f"Stay duration:  {travel.stay_duration} nights")
print(f"Python type:    {type(travel).__name__}  (not a string!)")
````

**Observations:**
- **Method 1 (prompt-only):** the model returns JSON-looking text, sometimes wrapped in a ```json code block. The values follow the text: `start_date` is "June 12th" and `stay_duration` can be "5 nights" as a string. The program has to clean the text and parse it with `json.loads`, and any change in format breaks it.
- **Method 2 (schema-enforced):** the API returns an object that matches the schema: `destination` is "Tokyo", `start_date` is in YYYY-MM-DD format and `stay_duration` is the integer 5. The text has no year, so the model has to assume one.

**Key insight:** method 2 is more reliable for a computer program. The field names and data types are guaranteed by the API, not by the model following instructions, so the result can be used directly in code without extra checks.
