Write and test a function named 'cos_similarity_calculation' that determines the semantic similarity between the sentences in your list and any text sentence using their vector representations and the cosine distance as a similarity measure.  

```python
# place your code here
def cos_similarity_calculation(sentences_list, query_sentence):
    list_embeddings = model.encode(sentences_list)
    query_embeddings = model.encode(query_sentence)

    similarity_scores = util.cos_sim(query_embeddings, list_embeddings)

    for sentence, score in zip(sentences_list, similarity_scores[0]):
        print(f"Score: {score:.4f} | Sentence: {sentence}")

    return similarity_scores

cos_similarity_calculation(sentences_list, "Doctors use technology to detect diseases early.")
```

Create a function that determines the cosine similarity between a vector and a batch of vectors using the cosine distance formula and the numpy library. Add code to demonstrate how to use this function.

```python
import numpy as np

def numpy_cosine_similarity(vector, batch_vectors):
    dot_products = np.dot(batch_vectors, vector)

    vector_norm = np.linalg.norm(vector)
    batch_norms = np.linalg.norm(batch_vectors, axis=1)

    cosine_similarities = dot_products / (vector_norm * batch_norms)

    return cosine_similarities

query_vector = model.encode("Doctors use technology to detect diseases early.")
batch_vectors = model.encode(sentences_list)

scores = numpy_cosine_similarity(query_vector, batch_vectors)

for sentence, score in zip(sentences_list, scores):
    print(f"Score: {score:.4f}\n Sentence:{sentence}")
```
