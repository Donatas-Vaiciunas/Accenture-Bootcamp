Initialize a client

```python
from dotenv import load_dotenv
import os
from pinecone import Pinecone

load_dotenv()
pc = Pinecone(api_key=os.environ["PINECONE_API_KEY"])
```

Create an index

```python
index_name = "index0"

if not pc.has_index(index_name):
    pc.create_index_for_model(
        name=index_name,
        cloud="aws",
        region="us-east-1",
        embed={
            "model": "llama-text-embed-v2",
            "field_map": {"text": "chunk_text"}
        }
    )

index = pc.Index(index_name)
```

Upser Data

```python
records = [
    {"_id": "rec1", "chunk_text": "Artificial intelligence helps doctors analyze medical images faster.", "category": "technology"},
    {"_id": "rec2", "chunk_text": "The weather is sunny today with a gentle morning breeze.", "category": "weather"},
    {"_id": "rec3", "chunk_text": "Vector databases allow you to search information by semantic meaning.", "category": "databases"},
    {"_id": "rec4", "chunk_text": "I love cooking homemade pasta with fresh tomato sauce and basil.", "category": "food"}
]

index.upsert_records(namespace="documents", records=records)
```

When you perform "Upsert data", add a list of your own text strings.

```python
my_records = [
    {"_id": "my1", "chunk_text": "The Eiffel Tower in Paris was completed in 1889."},
    {"_id": "my2", "chunk_text": "Python is a popular programming language for data science."},
    {"_id": "my3", "chunk_text": "Regular exercise improves heart health and mood."},
    {"_id": "my4", "chunk_text": "The football match ended in a dramatic penalty shootout."},
    {"_id": "my5", "chunk_text": "Electric cars are becoming cheaper and more common every year."},
]

index.upsert_records(namespace="documents", records=my_records)
```

```python
results = index.search(
    namespace="documents",
    query={"top_k": 2, "inputs": {"text": "Which coding language is good for analyzing data?"}},
)

for hit in results.result.hits:
    print(hit.id, round(hit.score, 3), hit.fields["chunk_text"])
```
